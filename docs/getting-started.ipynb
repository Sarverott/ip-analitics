```python
import paramiko
import pandas
import pprint

output = {"pwd":None,"whoami":None,"hostname":None,"ls -a":None,"python --version":None}

client = paramiko.client.SSHClient()
client.load_system_host_keys()
client.connect(input("ssh host:"), username=input("ssh user:"), password=input("ssh pass:"))
for commandline in output:
    stdin, stdout, stderr = client.exec_command(commandline)
    output[commandline]=stdout.read()
client.close()
pprint.pprint(output)
```

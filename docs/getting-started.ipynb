```python
import re
import pprint


commandlines = set()

commandUniq = set()

storyrows = list()

errorful_files = dict()

ttystorylines = glob.glob(os.path.join(os.environ["HOME"], './ttystory-*-*.txt'))

for version_of_tty in ttystorylines:
    with open(version_of_tty) as filestory:
        line_of_file = 0
        for sampleline in filestory.readlines():
            line_of_file +=1
            if not re.match(r"^\s+\d+[\*]?\s+", sampleline):
                delocalizedpath = "".join(unlocal_paths([version_of_tty]))
                if not delocalizedpath in errorful_files:
                    errorful_files[delocalizedpath] = dict()
                errorful_files[delocalizedpath][line_of_file]=sampleline
                # print("ERROR", , )
            else:
                commandlines.add(sampleline.strip())#.split(" ")[0]
                storyrows.append(sampleline.strip())
                commandUniq.add(" ".join(sampleline.strip().split(" ")[1:]))

print(len(commandlines))
print(len(storyrows))
print(len(commandUniq))
pprint.pprint(errorful_files)
print(list(commandUniq)[-1])
```

```python
pattern = r"^(?:(?:25[0-5]|2[0-4][0-9]|[01]?[0-9][0-9]?)\.){3}(?:25[0-5]|2[0-4][0-9]|[01]?[0-9][0-9]?)$"
# pattern regexp from: https://share.google/aimode/pmHeO91gux6z8t7Ge
print(
    {
        "valid":re.match(pattern, "1.1.1.1") is not None,
        "invalid":re.match(pattern, "1.1.f.1") is not None
    }
)
    
```

```python
optimized_pattern = r"^(?:(?:25[0-5]|2[0-4]/d|[01]?/d/d?)\.){3}(?:25[0-5]|2[0-4]/d|[01]?/d/d?)$"

print(
    {
        "valid":re.match(optimized_pattern, "1.1.1.1") is not None,
        "invalid":re.match(optimized_pattern, "1.1.f.1") is not None
    }
)
    
```

```python
even_more_optimized_pattern = r"^((2(5[0-5]|[0-4]\d)|1?\d+)\.){3}(2(5[0-5]|[0-4]\d)|1?\d+)$"

print(
    {
        "valid":re.match(even_more_optimized_pattern, "0.1.1.0") is not None,
        "invalid":re.match(even_more_optimized_pattern, "1.1111.0.0") is not None
    }
)
    
```

```python
even_more_optimized_pattern = r"^((2(5[0-5]|[0-4]\d)|1?\d?\d)\.){3}(2(5[0-5]|[0-4]\d)|1?\d?\d)$"



print(
    {
        "valid":re.match(even_more_optimized_pattern, "0.1.1.0") is not None,
        "invalid":re.match(even_more_optimized_pattern, "1..0.0") is not None
    }
)
print(
    {
        "valid":re.match(even_more_optimized_pattern, "0.1.1.0") is not None,
        "invalid":re.match(even_more_optimized_pattern, "1.1111.0.0") is not None
    }
)
print(
    {
        "valid":re.match(even_more_optimized_pattern, "255.255.255.255") is not None,
        "invalid":re.match(even_more_optimized_pattern, "256.1.0.0") is not None
    }
)
```

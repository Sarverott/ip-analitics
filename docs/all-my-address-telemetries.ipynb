```python
import requests
import urllib
import json

api_list = [
    "https://api.myip.com/",
    "https://ipapi.co/json/",
    "https://api.ipify.org/?format=json",
    "https://api.my-ip.io/v2/ip.json",
    "https://httpbin.org/get"
]

output=dict()

for api_url in api_list:
    output[urllib.parse.urlsplit(api_url).netloc]=[api_url,list()]
    try:
        req = requests.get(api_url)
        output[urllib.parse.urlsplit(api_url).netloc][1].append(
            req.status_code
        )
        output[urllib.parse.urlsplit(api_url).netloc][1].append(
            req.text
        )
        output[urllib.parse.urlsplit(api_url).netloc][1].append(
            dict(req.headers)
        )
        output[urllib.parse.urlsplit(api_url).netloc][1].append(
            dict(req.json())
        )
    except Exception as error:
        output[urllib.parse.urlsplit(api_url).netloc].append(str(error))

#import ast

#print(ast.dump(output, indent=4))
print([[queryData, output[queryData][0],  output[queryData][1][0]] for queryData in output])
# for keeping sensitive data this output should be not presented in notebook's printing
```
